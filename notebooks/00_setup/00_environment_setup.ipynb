# 00. 環境セットアップとログイン確認

## この章のゴール
W&B にログインできることを確認し、最初の run を 1 つ作ります。

## 注目する W&B コード
```python
with wandb.init(project=PROJECT) as run:
    run.log({"check/ok": 1})
```

## 接続先を選ぶ

標準は Multi-tenant SaaS です。Colab Secrets またはローカルの `.env` に `WANDB_API_KEY` を設定してください。

Dedicated Cloud を使う場合は、同じ場所に `WANDB_BASE_URL=https://your-org.wandb.io` を設定し、その接続先で発行した API key を使います。URL に project のパスは付けません。SaaS なら `WANDB_BASE_URL` は未設定または空欄にします。

`WANDB_ENTITY`（チーム）と `WANDB_PROJECT` は任意です。Colab では登録した Secrets への Notebook アクセスを許可してください。


````python
# リポジトリ内ではルートを使い、Colab 単体では共通コードを取得します。
from pathlib import Path
import os
import sys
import urllib.request

root = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "course_utils.py").is_file()), Path.cwd())
if not (root / "course_utils.py").exists():
    urllib.request.urlretrieve(
        "https://raw.githubusercontent.com/nejumi/models-initial-course/main/course_utils.py",
        root / "course_utils.py",
    )
os.chdir(root)
sys.path.insert(0, str(root))

from course_utils import setup_wandb
PROJECT, ENTITY, WANDB_INIT = setup_wandb()
````

````python
import wandb

with wandb.init(**WANDB_INIT, name="00-login-check", tags=["course", "setup"]) as run:
    run.log({"check/ok": 1})
    run_url = run.url

print("Run URL:", run_url)
````

## W&B App で確認すること
- `00-login-check` という run が作成されている
- `check/ok` という metric が 1 点記録されている
- Project と Entity が想定どおりになっている
